simple if-else statements

```python
def factorial_value(num):
  factorial =1

  if num== 0:
    return factorial

  else:
    for i in range (1,num+1):  #range(1,number+1)include 1 and exclude 6
       factorial= factorial *i    #1*2*3*4*5
  return factorial

print(factorial_value(5))
```

FACtorial of a number ---> IS PRODUCT of all the positive integers less than or equal to number

```python
#factorial of 5--- 5x4x3x2x1 =120

number= int(input("enter a number to fibd its factorial: "))

factorial = 1

if number == 0:
  print("The factorial of 0 is 1")

else:
  for i in range (1,number+1): #range(1,number+1)include 1 and exclude 6
    factorial= factorial *i    #1*2*3*4*5
  print("The factorial of ", number, "is", factorial)
```
