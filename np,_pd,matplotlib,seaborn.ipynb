time taken by a list

```python
from time import process_time

python_list = [i for i in range(10000)]

start_time = process_time()

python_list= [1+5 for i in python_list]

end_time= process_time()

print(end_time-start_time)
```

```python
from time import process_time
import numpy as np # Ensure numpy is imported

np_array = np.arange(10000)

start_time = process_time()

np_array += 5

end_time = process_time()

print(end_time-start_time)
```

Initial placeholders in numpy arrays


```python
# create a numpy array of zeroes

x= np.zeros((4,5)) #shape of arrays
print(x)
```

```python
#creating numpy arrays of ones

y= np.ones((3,3))
print(y)
```

```python
#creating of array of particular values

z= np.full((5,4),5)
print(z)  #4x5 array with all the values as 5
```

```python
#creating identity matrix

v= np.eye(3) #only 4x4 or 5x5 will work
print(v)
```

```python
import numpy as np
#create a numpy array with random values (gives value between 0 and 1)

b= np.random.random((3,4))
print(b)
```

```python
#random integer values array within a specific range

import numpy as np
c= np.random.randint(10, 100, size=(3,5)) # specific range (low, high), matrix dimension
print(c)
```

```python
#create an array of evenly spaced values
# evenly spaced ---- specifying the no. of values

d= np.linspace(10,30,5) #start, end, number of values
print(d)
```

```python
# array of evenly spaced values ---- specifying the steps
#between particular range

e= np.arange(10,30,5) #start, end, step
print(e)
```

```python
#convert a list to a numpy array

list2 = [10,20,30,40,50]

np_array= np.asarray(list2)
print(np_array)
type(np_array)
```
